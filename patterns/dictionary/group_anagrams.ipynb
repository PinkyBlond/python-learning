```python
def group_anagrams(words):
   result= {}
   for i in words:
      anagram_key = "".join(sorted(i))
      if  anagram_key in result:
          result[anagram_key].append(i)
      else:
          result[anagram_key]=[i]
   return list(result.values())
words = ["eat", "tea", "tan", "ate", "nat", "bat"]
group_anagrams(words)
```
