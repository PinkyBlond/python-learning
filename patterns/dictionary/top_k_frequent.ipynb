```python
nums = [1, 1, 1, 2, 2, 3]
k = 2

def top_k_frequent(nums, k):
    stats = {}
    for i in nums:
        stats[i]= stats.get(i,0)+1    
    result=[]
    for _ in range(k):
        max_key = 0
        max_count = 0
        for i in stats:
            if stats[i]>max_count:
                max_count=stats[i]
                max_key = i
        result.append(max_key)
        del stats[max_key]
    return result

print(top_k_frequent([1, 1, 1, 2, 2, 3], 2))
print(top_k_frequent([4, 4, 4, 5, 5, 6, 6, 6, 6], 1))
print(top_k_frequent([1], 1))
print(top_k_frequent([], 0))
```
