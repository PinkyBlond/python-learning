```python
def longest_consecutive(nums):
   unique_set = set(nums)
   num = 0 
   current_length=0
   max_length=0 
   for num in unique_set:
     if (num-1 not in unique_set):
            current_length = 1
            max_length = max(max_length, current_length)
            next_num = num+1 
            while next_num in unique_set:
                current_length+=1
                next_num+=1
                max_length = max(max_length, current_length)
   return max_length
    
print(longest_consecutive([1,2,3, 1,2,100,101,102,103,45]))
```
