```python
def is_palindrome(s):
    cleaned =[]
    for ch in s:
       if ch.isalnum():
        cleaned.append(ch.lower())
    left =0
    right= len(cleaned)-1
    while right>left:
        if cleaned[left]==cleaned[right]:
            left+=1
            right-=1
            continue   
        else:
          return False
    return True    


print('level',is_palindrome('level'))
print('A man, a plan, a canal: Panama', is_palindrome('A man, a plan, a canal: Panama'))
```
