```python
def is_valid(s):
    if s=="":
        return True
    pairs = {
        ")": "(",
        "]": "[",
        "}": "{"
    }
    stack=[]    
    for ch in (s):
           if (ch not in pairs):
                stack.append(ch)
           elif (ch in pairs): 
               if  (not stack):
                   return False
               else:
                 if (pairs[ch]==stack.pop()) :
                    continue
                 else:
                     return False
    return len(stack) == 0

print(is_valid(""))          # True
print(is_valid("())"))          # False
print(is_valid("()[]{}"))      # True
print(is_valid("{[]}"))        # True
print(is_valid("(]"))          # False
print(is_valid("([)]"))      # False
print(is_valid("((("))        # False
print(is_valid(")"))          # False
```
