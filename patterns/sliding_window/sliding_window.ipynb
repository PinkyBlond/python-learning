```python
def the_longest_substring_length_wihout_repeat(str):
  longest = 0
  if str:
   substr = {}
   max_sub_str=0
   for ch in str:
       if ch in substr.keys():
          max_sub_str = max(len(substr.items()),max_sub_str)
          substr={}
          substr[ch]=1
       else:
          substr[ch]=1
          max_sub_str = max(len(substr.items()),max_sub_str) 
   return  max_sub_str
  else:
    return longest

print(the_longest_substring_length_wihout_repeat("tmmzuxt"))
```

```python
def longest_substring(s):
    seen = {}
    max_length = 0
    start = 0
    
    for end, ch in enumerate(s):
        if ch in seen and seen[ch] >= start:
            start = seen[ch] + 1
        seen[ch] = end
        max_length = max(max_length, end - start + 1)
    
    return max_length
print(longest_substring("tmmzuxt"))
```
