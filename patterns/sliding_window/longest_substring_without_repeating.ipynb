```python
def longest_substring_without_repeating(s):
    last_seen={}
    longest_substring=0
    start=0
    end=0
    print(s)
    for index, ch in enumerate(s):
        print(ch)
        if ch not in last_seen:
            longest_substring=max(longest_substring, end-start)
            print(ch, last_seen,"start",start, "end",end, longest_substring)
        else: 
              longest_substring=max(longest_substring, end-start)
              if last_seen[ch]+1>start:
                  start=last_seen[ch]+1
              print(ch,last_seen,"start",start, "end",end, longest_substring)
        last_seen[ch]=index
        end = index+1
        longest_substring=max(longest_substring, end-start)
    return longest_substring

print(longest_substring_without_repeating("")," -> 0")
print(longest_substring_without_repeating("tmmzuxt")," -> 5")
print(longest_substring_without_repeating("pwwkew")," -> 3")
print(longest_substring_without_repeating("abba")," -> 2")
print(longest_substring_without_repeating("dadvaddaf")," -> 3")
print(longest_substring_without_repeating("adftg")," -> 5")
```
