```python
#Return the longest common prefix shared by all strings in the list.
# my brut force solution
def longest_common_prefix(strs):
    if strs:
        common_prefix = []
        last_prefix=[]
        for n_letter in range(0, len(strs[0])):
            for index, word in enumerate(strs):
               # if word is not empty string 
               if word and  n_letter<=len(word)-1:
                   #if this is the first word we are taking this letter into prefix
                   if index==0:
                       common_prefix.append(word[n_letter])
                   # if this is the last word
                   elif index == len(strs)-1: 
                     # and letter is in common prefix  - we update last prefix
                     if word[n_letter] == common_prefix[n_letter]: 
                         last_prefix.append(word[n_letter])
                     elif word[n_letter] != common_prefix[n_letter]:
                         return "".join(last_prefix)
                   # if letter is not in common prefix - we have to return last prefix
                   elif word[n_letter] != common_prefix[n_letter]:  
                           return "".join(last_prefix)
               else:
                      return "".join(last_prefix)
        return "".join(last_prefix)
    else:
        return ''

        


print(longest_common_prefix(["flower", "flow", "flight"]), "fl")
print(longest_common_prefix(["dog", "racecar", "car"]), "")
print(longest_common_prefix(["interspecies", "interstellar", "interstate"]), "inters")
print(longest_common_prefix(["a"]), "a")
print(longest_common_prefix([""]), "")
print(longest_common_prefix([]), "")
```
