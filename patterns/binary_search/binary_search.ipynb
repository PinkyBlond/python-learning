```python
def binary_search(nums, target):
    left = 0
    right = len(nums) - 1

    while left <= right:
        middle = (right + left) //2
        if nums[middle] == target:
            return middle
        elif nums[middle] < target:
            left=middle +1 
        else:
            right = middle-1
    return -1
    
print(binary_search([1, 3, 5, 7, 9], 5))
print(binary_search([1, 3, 5, 7, 9], 8))
```
