```python
def contains_duplicate(nums):
    unique = set(nums)
    return (len(unique) != len(nums))
        


print(contains_duplicate([1,2,3,1]), "  → True")
print(contains_duplicate([1,2,3,4]), "  → False")
print(contains_duplicate([1,1,1,3]), "  → True")
print(contains_duplicate([]), "  → False")
print(contains_duplicate([]), "  → False")

```
