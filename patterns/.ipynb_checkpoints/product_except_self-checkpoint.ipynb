```python
def product_except_self(nums):
    product= [1] * len(nums)
    for i  in range (0, len(nums)): 
        for j  in range (0, len(nums)):  
          if i!=j:
            product[i] = product[i]*nums[j]
    return product

        
print(product_except_self([1,2,3,4]),"[24,12,8,6]")
```

```python
def product_except_self(nums):
    final_product= [1] * len(nums)
    all_to_the_left=[]
    all_to_the_right=[]
    for index, num  in enumerate(nums): 
         all_to_the_left=nums[0:index]
         all_to_the_right=nums[index+1:]
         product= []
         product.extend(all_to_the_left)
         product.extend(all_to_the_right)
         for pnum in (product):
           final_product[index]=final_product[index]*pnum
    return final_product

        
print(product_except_self([1,2,3,4]),"[24,12,8,6]")
```

```python
# Pass 1 → fill left products.
# Pass 2 → add right products.

def product_except_self(nums):
    final_product= [1] * len(nums)
    all_to_the_left=1
    all_to_the_right=1
    for index, num  in enumerate(nums):  
        final_product[index] *= all_to_the_left
        all_to_the_left *= nums[index]
    for index  in range(len(nums)-1,-1,-1):
        final_product[index]*= all_to_the_right
        all_to_the_right *=nums[index]
    return final_product


        
print(product_except_self([1,2,3,4]),"[24,12,8,6]")
```
