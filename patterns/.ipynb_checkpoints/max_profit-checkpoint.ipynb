You are given a list where:

prices[i]

is the stock price on day i.

You may:

buy once
sell once
the buy must happen before the sell.

Return the maximum profit.

If no profit is possible, return 0.

```python
def max_profit(prices):
    profit =0
    for i in range(0, len(prices)-1):
        for j in range(i+1, len(prices)):
            profit =max(profit, prices[j]-prices[i])
    return profit


prices = [7, 6, 4, 3, 1]

print(max_profit(prices))
```

```python
#Use Lowest seen so far,	Best profit
def max_profit(prices):
    if not prices:
        return 0
    lowest_seen_so_far = prices[0]
    best_profit =0
    for price in (prices):
        lowest_seen_so_far = min(lowest_seen_so_far, price)
        best_profit =max(best_profit, price-lowest_seen_so_far)
    return best_profit


prices = [1]

print(max_profit(prices))
```
