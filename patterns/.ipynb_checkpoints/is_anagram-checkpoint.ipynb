```python
def is_anagram(s, t):
    s_word= {}
    t_word= {}
    for ch in (s):
        s_word[ch]=s_word.get(ch,0)+1
    for ch in (t):
        t_word[ch]=t_word.get(ch,0)+1
    for ch in (s_word):
        if ch not in (t_word):
            return False
        elif s_word[ch]!=t_word[ch]:
            return False 
    for ch in (t_word):
        if ch not in (s_word):
            return False
        elif t_word[ch]!=s_word[ch]:
            return False
    return True
    
print(is_anagram("anagram", "nagaram")," → True")
print(is_anagram("rat", "car")," → False")
print(is_anagram("", "")," → True")
```
