```python
def merge_sorted_arrays(nums1, nums2):
    nums_joined=[]
    pointer1=0
    pointer2=0
    still_to_go1=True
    still_to_go2=True
    if not nums1:
        return nums2
    if not nums2:
        return nums1
        
    while still_to_go1 and still_to_go2:
            if  nums1[pointer1]<nums2[pointer2]:
               nums_joined.append(nums1[pointer1])
               if pointer1 <len(nums1)-1:
                    pointer1+=1
               else:
                    nums_joined.extend(nums2[pointer2:])
                    still_to_go1= False
            elif nums1[pointer1]==nums2[pointer2]:
               nums_joined.append(nums1[pointer1])
               if pointer1 <len(nums1)-1:
                    pointer1+=1
               else:
                    nums_joined.extend(nums2[pointer2:])
                    still_to_go1= False
               nums_joined.append(nums2[pointer2])
               if pointer2 <len(nums2)-1:
                    pointer2+=1
               else:
                    nums_joined.extend(nums1[pointer1:])
                    still_to_go2= False
            else:
                nums_joined.append(nums2[pointer2])
                if pointer2 <len(nums2)-1:
                    pointer2+=1
                else:
                    nums_joined.extend(nums1[pointer1:])
                    still_to_go2= False
        
    return nums_joined

nums1 = [1,3,5]
nums2 = [2,4,6]

print(merge_sorted_arrays(nums1, nums2),"→ [1,2,3,4,5,6]")

nums1 = [1,2,2]
nums2 = [2,3]

print(merge_sorted_arrays(nums1, nums2),"→ [1,2,2,2,3]")


nums1 = []
nums2 = [1]
print(merge_sorted_arrays(nums1, nums2),"→ [1]")
```
