```python
sring_to_compress= "aabbbbcccccdde"
def compress_string(sring_to_compress):
    if sring_to_compress =="":
     return ""
    result=[sring_to_compress[0]]
    n=1
    current_letter = sring_to_compress[0]
    for letter in sring_to_compress[1:]:
        if letter== current_letter:
            n+=1
        else:
            result.append(str(n))
            n=1
            result.append(letter)
            current_letter = letter
    result.append(str(n))
    compressed_string="".join(result)
    if len(compressed_string)>=len(sring_to_compress):
        compressed_string= sring_to_compress
    return compressed_string
compress_string(sring_to_compress)
```
